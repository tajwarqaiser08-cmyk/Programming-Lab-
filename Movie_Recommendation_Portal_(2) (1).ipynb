# Movie Recommendation Portal



## Load movies from movies.txt

```python
# Each line in movies.txt:  title,year,genre,actor1,actor2
def load_movies():
    movies = []
    try:
        with open("movies.txt", "r") as file:
            for line in file:
                parts = line.strip().split(",")
                if len(parts) == 5:
                    movies.append(parts)
    except FileNotFoundError:
        print("movies.txt not found. Put it in the same folder.")
    except PermissionError:
        print("No permission to read movies.txt.")

    print("Loading finished.")
    return movies
```

## Register and login

```python
# Reads users.txt. Each line:  username,password,genre,actor
def read_users():
    users = []
    try:
        with open("users.txt", "r") as file:
            for line in file:
                parts = line.strip().split(",")
                if len(parts) == 4:
                    users.append(parts)
    except FileNotFoundError:
        pass
    return users


# Creates a new account and saves it in users.txt.
def register_user():
    username = input("Choose a username: ")
    password = input("Choose a password: ")
    genre = input("Favourite genre (e.g. Action, Comedy, Drama, Horror): ")
    actor = input("Favourite actor: ")
    for user in read_users():
        if user[0] == username:
            print("Username already taken.")
            return
    try:
        with open("users.txt", "a") as file:
            file.write(username + "," + password + "," + genre + "," + actor + "\n")
        print("Account created. Now log in.")
    except PermissionError:
        print("No permission to write to users.txt.")


# Checks username and password (3 tries).
# Returns the user's details if correct, otherwise an empty list.
def user_login():
    users = read_users()
    for attempt in range(3):
        username = input("Username: ")
        password = input("Password: ")
        for user in users:
            if user[0] == username and user[1] == password:
                print("Welcome,", username)
                return user
        print("Wrong username or password.")
    return []
```

## Show, recommend and search movies

```python
# Prints movies.
def show_movies(movie_list):
    if len(movie_list) == 0:
        print("No movies found.")
    for m in movie_list:
        print(m[0], "(" + m[1] + ") -", m[2], "-", m[3], "&", m[4])


# Shows movies that match the user's favourite genre and actor.
def recommend_movies(movies, genre, actor):
    found = []
    for m in movies:
        if m[2].lower() == genre.lower() and actor.lower() in (m[3] + m[4]).lower():
            found.append(m)
    show_movies(found)




# Returns movies that contain the keyword (title, year, genre or actor).
def search_record(movies, keyword):
    found = []
    for m in movies:
        if keyword.lower() in (m[0] + m[1] + m[2] + m[3] + m[4]).lower():
            found.append(m)
    return found
```

## Add a movie

```python
# Adds a new movie to movies.txt.
def add_record(movies):
    title = input("Title: ")
    try:
        year = int(input("Year: "))
    except ValueError:
        print("Year must be a number.")
        return
    genre = input("Genre: ")
    actor1 = input("First actor: ")
    actor2 = input("Second actor: ")
    try:
        with open("movies.txt", "a") as file:
            file.write(f"{title},{year},{genre},{actor1},{actor2}\n")
        movies.append([title, str(year), genre, actor1, actor2])
        print("Movie added.")
    except PermissionError:
        print("No permission to write to movies.txt.")
```

**Movie Recommendation **

```python
def recommend_movie():
    """Recommend movies according to user's selected genre."""

    print("\n===== MOVIE RECOMMENDATION =====")

    genre = input(
        "Choose a genre (Sci-Fi, Romance, Horror, Action, Animation): "
    ).strip().lower()

    found = False

    try:
        with open("movies.txt", "r") as file:

            print("\nRecommended Movies for", genre.title())

            for line in file:

                details = line.strip().split(",")

                if len(details) != 4:
                    continue

                name = details[0]
                movie_genre = details[1]
                rating = details[2]
                year = details[3]

                if movie_genre.lower() == genre:

                    print("\nMovie Name:", name)
                    print("Genre:", movie_genre)
                    print("Rating:", rating, "/10")
                    print("Release Year:", year)

                    found = True

            if not found:
                print("Sorry! No movies found for this genre.")

    except FileNotFoundError:
        print("Error: movies.txt file was not found.")

    except PermissionError:
        print("Error: Cannot access movies.txt file.")
```

## Generate report

```python
# Prints a small report and saves it in report.txt.
def generate_report(movies):
    lines = ["Total movies: " + str(len(movies)),
             "Registered users: " + str(len(read_users()))]
    try:
        with open("report.txt", "w") as file:
            for line in lines:
                print(line)
                file.write(line + "\n")
    except PermissionError:
        print("No permission to write to report.txt.")
```

## Main menu loop (run this cell to start the portal)

```python
def main():
    movies = load_movies()

    user = []
    while len(user) == 0:
        print("\n1. Register\n2. Login\n3. Exit")
        choice = input("Choose: ")
        if choice == "1":
            register_user()
        elif choice == "2":
            user = user_login()
        elif choice == "3":
            print("Goodbye!")
            return
        else:
            print("Invalid choice.")

    while True:
        print("\n1. My recommendations\n2. Search\n3. Add a movie\n4. Report\n5. Exit")
        choice = input("Choose: ")
        if choice == "1":
            recommend_movies(movies, user[2], user[3])
        elif choice == "2":
            show_movies(search_record(movies, input("Search: ")))
        elif choice == "3":
            add_record(movies)
        elif choice == "4":
            generate_report(movies)
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid choice.")


main()
```
